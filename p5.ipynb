```python
import re
```

```python
import numpy as np
from tensorflow import keras
```

```python
keras.utils.set_random_seed(42)
epochs = 200
text = """We are about to study the idea of a computational process.
Computational processes are abstract beings that inhabit computers.
As they evolve, processes manipulate other abstract things called data.
The evolution of a process is directed by a pattern of rules
called a program. People create programs to direct processes. In effect,
we conjure the spirits of the computer with our spells."""
words = re.findall(r"[a-z]{2,}", text.lower())
vocabulary = sorted(set(words))
word_to_id = {word: index for index, word in enumerate(vocabulary)}
tokens = [word_to_id[word] for word in words]
contexts = np.array([tokens[i - 2:i] + tokens[i + 1:i + 3] for i in range(2, len(tokens) - 2)])
targets = np.array(tokens[2:-2])
```

```python
model = keras.Sequential([
    keras.Input(shape=(4,), dtype="int32"),
    keras.layers.Embedding(len(vocabulary), 10),
    keras.layers.GlobalAveragePooling1D(),
    keras.layers.Dense(len(vocabulary), activation="softmax"),
])
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.01),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
model.fit(contexts, targets, epochs=epochs, batch_size=8, verbose=0)
```

```python
loss, accuracy = model.evaluate(contexts, targets, verbose=0)
print(f"Training loss: {loss:.4f}, training accuracy: {accuracy:.4f}")
context = ["we", "are", "to", "study"]
prediction = model.predict(np.array([[word_to_id[word] for word in context]]), verbose=0)
print(f"Context: {context}, predicted word: {vocabulary[prediction.argmax()]}")
```
